```python
import pandas as pd
from sqlalchemy import create_engine, text
import warnings
warnings.filterwarnings('ignore')

# Connect to database
engine = create_engine('postgresql://analytics_user:analytics_password_123@localhost:5432/ecommerce_analytics')
print('✅ Connected to PostgreSQL!')
```

```python
print("=" * 80)
print("🎉 PHASE 4 COMPLETE: DATA WAREHOUSE FULLY LOADED!")
print("=" * 80)

summary_df = pd.read_sql("""
SELECT 
    'Staging Tables' AS category, COUNT(*) AS count FROM information_schema.tables WHERE table_schema = 'staging'
UNION ALL SELECT 'Dimension Tables', COUNT(*) FROM information_schema.tables WHERE table_schema = 'dimensions'
UNION ALL SELECT 'Fact Tables', COUNT(*) FROM information_schema.tables WHERE table_schema = 'facts'
UNION ALL SELECT 'Analytics Views', COUNT(*) FROM pg_views WHERE schemaname = 'analytics';
""", engine)

print("\n📊 DATABASE STRUCTURE:")
print(summary_df.to_string(index=False))

rows_df = pd.read_sql("""
SELECT 'Staging' AS layer, SUM(n_live_tup)::BIGINT AS total_rows FROM pg_stat_user_tables WHERE schemaname = 'staging'
UNION ALL SELECT 'Dimensions', SUM(n_live_tup)::BIGINT FROM pg_stat_user_tables WHERE schemaname = 'dimensions'
UNION ALL SELECT 'Facts', SUM(n_live_tup)::BIGINT FROM pg_stat_user_tables WHERE schemaname = 'facts';
""", engine)

print("\n📈 DATA VOLUME:")
print(rows_df.to_string(index=False))

print(f"\n✅ PHASE 4: 3.1M+ ROWS, 5 ANALYTICS VIEWS CREATED! 🎉")
```

```python
top_products = pd.read_sql("""
SELECT 
    product_category_name,
    orders_sold,
    ROUND(gmv::NUMERIC, 2) AS gmv,
    ROUND(avg_review_score, 2) AS rating
FROM analytics.mart_product_performance
WHERE gmv IS NOT NULL
ORDER BY gmv DESC
LIMIT 10;
""", engine)

print("\n📊 TOP 10 PRODUCTS BY GMV:")
print(top_products.to_string(index=False))
```

```python
monthly_sales = pd.read_sql("""
SELECT 
    year_month,
    order_count,
    ROUND(total_revenue::NUMERIC, 2) AS revenue,
    ROUND(avg_order_value::NUMERIC, 2) AS avg_order_value,
    ROUND(on_time_pct, 2) AS on_time_pct
FROM analytics.mart_sales_summary
ORDER BY year_month;
""", engine)

print("\n📈 MONTHLY SALES TREND:")
print(monthly_sales.to_string(index=False))
print(f"\nTotal months: {len(monthly_sales)}")
```

```python
top_customers = pd.read_sql("""
SELECT 
    customer_state,
    lifetime_orders,
    ROUND(lifetime_value::NUMERIC, 2) AS lifetime_value,
    ROUND(avg_review_score, 2) AS avg_rating
FROM analytics.mart_customer_analysis
WHERE lifetime_value IS NOT NULL
ORDER BY lifetime_value DESC
LIMIT 10;
""", engine)

print("\n💰 TOP 10 CUSTOMERS BY LIFETIME VALUE:")
print(top_customers.to_string(index=False))
```
